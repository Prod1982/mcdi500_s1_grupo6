# Fase 2 - Pipeline de obtención, limpieza y transformación de datos

Este notebook desarrolla el pipeline inicial de la Fase 2 del proyecto.  
El objetivo es cargar el dataset AI Job Impact, explorar su estructura, limpiar registros, transformar variables básicas, validar el resultado y exportar una versión procesada para fases posteriores.

```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_RAW = PROJECT_ROOT / "data" / "raw" / "ai_job_impact.csv"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed" / "ai_job_impact_clean.csv"
RESULTS_VALIDATION = PROJECT_ROOT / "results" / "resumen_validacion_f2.txt"
RESULTS_STATS = PROJECT_ROOT / "results" / "estadisticas_dataset.txt"

SRC_PATH = PROJECT_ROOT / "src"
sys.path.append(str(SRC_PATH))

print("Ruta del proyecto:", PROJECT_ROOT)
print("Dataset original:", DATA_RAW)
print("Dataset procesado:", DATA_PROCESSED)
```

```python
from preprocessing import (
    cargar_datos,
    explorar_datos,
    limpiar_datos,
    transformar_datos,
    validar_datos,
    exportar_resultados
)
```

```python
df_original = cargar_datos(DATA_RAW)

df_original.head()
```

```python
resumen_inicial = explorar_datos(df_original)
resumen_inicial
```

```python
df_original.info()
```

```python
df_original.isna().sum()
```

```python
df_original.duplicated().sum()
```

```python
df_original.describe(include="all")
```

```python
df_limpio = limpiar_datos(df_original)

print("Dimensiones originales:", df_original.shape)
print("Dimensiones después de limpieza:", df_limpio.shape)
print("Duplicados después de limpieza:", df_limpio.duplicated().sum())
```

```python
df_transformado = transformar_datos(df_limpio)

df_transformado.head()
```

```python
validaciones = validar_datos(df_transformado)
validaciones
```

```python
ruta_exportada = exportar_resultados(df_transformado, DATA_PROCESSED)

print("Dataset procesado exportado en:", ruta_exportada)
```

```python
RESULTS_VALIDATION.parent.mkdir(parents=True, exist_ok=True)

with open(RESULTS_VALIDATION, "w", encoding="utf-8") as archivo:
    archivo.write("Resumen de validación Fase 2\n")
    archivo.write("============================\n\n")
    for clave, valor in validaciones.items():
        archivo.write(f"{clave}: {valor}\n")

print("Resumen de validación guardado en:", RESULTS_VALIDATION)
```

```python
estadisticas = df_transformado.describe(include="all").to_string()

with open(RESULTS_STATS, "w", encoding="utf-8") as archivo:
    archivo.write("Estadísticas descriptivas del dataset procesado\n")
    archivo.write("==============================================\n\n")
    archivo.write(estadisticas)

print("Estadísticas guardadas en:", RESULTS_STATS)
```

```python
print("Pipeline Fase 2 ejecutado correctamente.")
print("Archivo procesado:", DATA_PROCESSED.exists())
print("Resumen de validación:", RESULTS_VALIDATION.exists())
print("Estadísticas:", RESULTS_STATS.exists())
```
